1. (CREATE / INSERT) Experimento Comercial

```python
%sql
CREATE OR REPLACE TABLE cursodataacademy.proyecto_dataandino.tb_reservas_prueba AS
SELECT * 
FROM cursodataacademy.proyecto_dataandino.hotel_bookings 
WHERE hotel = 'City Hotel' AND is_canceled = 1;
```

2. (UPDATE) Limpieza de error de bebés

```python
%sql
UPDATE cursodataacademy.proyecto_dataandino.tb_reservas_prueba
SET babies = 0;
```

3. (CAST / CONCAT) Formato de fecha legible

```python
%sql
SELECT 
    CONCAT(
        CAST(arrival_date_day_of_month AS STRING), '-', 
        arrival_date_month, '-', 
        CAST(arrival_date_year AS STRING)
    ) AS Fecha_Llegada
FROM cursodataacademy.proyecto_dataandino.hotel_bookings;
```

4. (LIKE) Filtro de canal específico

```python
%sql
SELECT *
FROM cursodataacademy.proyecto_dataandino.hotel_bookings
WHERE distribution_channel LIKE 'T%';
```

5. (COUNT / SUM / AVG) Métricas globales

```python
%sql
SELECT 
    COUNT(*) AS Cantidad_Reservas,
    SUM(total_of_special_requests) AS Total_Peticiones,
    AVG(adr) AS Tarifa_Promedio
FROM cursodataacademy.proyecto_dataandino.hotel_bookings;
```

6. (DELETE) Borrado de reservas de prueba

```python
%sql
DELETE FROM cursodataacademy.proyecto_dataandino.tb_reservas_prueba
WHERE adr = 0.0;
```

7. (GROUP BY) Volumen por canal

```python
%sql
SELECT 
    distribution_channel,
    COUNT(*) AS Total_Reservas
FROM cursodataacademy.proyecto_dataandino.hotel_bookings
GROUP BY distribution_channel;
```

8. (CASE WHEN) Clasificación de precios

```python
%sql
SELECT 
    adr,
    CASE 
        WHEN adr < 50 THEN 'Económico'
        WHEN adr BETWEEN 50 AND 150 THEN 'Estándar'
        ELSE 'Premium'
    END AS Segmento_Precio
FROM cursodataacademy.proyecto_dataandino.hotel_bookings;
```

9. (Funciones de Fecha) Booking Date

```python
%sql
WITH ReservasConFechas AS (
    SELECT 
        lead_time,
        TO_DATE(CONCAT(arrival_date_year, '-', arrival_date_month, '-', arrival_date_day_of_month), 'yyyy-MMMM-d') AS arrival_date_real
    FROM cursodataacademy.proyecto_dataandino.hotel_bookings
)
SELECT 
    arrival_date_real AS Fecha_Llegada,
    lead_time,
    DATE_SUB(arrival_date_real, CAST(lead_time AS INT)) AS Fecha_Reserva
FROM ReservasConFechas;
```

10. (Funciones Matemáticas) Cuota de Riesgo

```python
%sql
SELECT 
    lead_time,
    CEIL(lead_time / 30.0) AS Meses_Anticipacion
FROM cursodataacademy.proyecto_dataandino.hotel_bookings;
```
