Hipótesis 1: ¿Los hijos aseguran el viaje?

```python
%sql
SELECT 
    CASE 
        WHEN children > 0 OR babies > 0 THEN 'Con Hijos'
        ELSE 'Sin Hijos'
    END AS Tipo_Familia,
    AVG(is_canceled) AS Tasa_Caidas
FROM cursodataacademy.proyecto_dataandino.hotel_bookings
GROUP BY 
    CASE 
        WHEN children > 0 OR babies > 0 THEN 'Con Hijos'
        ELSE 'Sin Hijos'
    END;
```

Conclusión: Ambos grupos muestran una tasa de caída altísima. Viajar con niños no garantiza mayor compromiso, por lo que la composición familiar no sirve para prever ingresos fijos.

Hipótesis 2: El impacto de las solicitudes especiales

```python
%sql
SELECT 
    total_of_special_requests,
    AVG(is_canceled) AS Probabilidad_Cancelacion
FROM cursodataacademy.proyecto_dataandino.hotel_bookings
GROUP BY total_of_special_requests
ORDER BY total_of_special_requests ASC;
```

Conclusión: A mayor cantidad de peticiones especiales, la tasa de cancelación se desploma drásticamente. La interacción y personalización consolidan fuertemente el viaje.

Hipótesis 3: ¿Cambiar la reserva es señal de compromiso?

```python
%sql
SELECT 
    CASE 
        WHEN booking_changes = 0 THEN 'Cero Modificaciones'
        WHEN booking_changes BETWEEN 1 AND 3 THEN '1 a 3 Modificaciones'
        ELSE 'Más de 3 Modificaciones'
    END AS Frecuencia_Cambios,
    AVG(is_canceled) AS Tasa_Caidas
FROM cursodataacademy.proyecto_dataandino.hotel_bookings
GROUP BY 
    CASE 
        WHEN booking_changes = 0 THEN 'Cero Modificaciones'
        WHEN booking_changes BETWEEN 1 AND 3 THEN '1 a 3 Modificaciones'
        ELSE 'Más de 3 Modificaciones'
    END;
```

Conclusión: Quienes jamás modifican su reserva cancelan en un 40% de los casos. Por el contrario, interactuar modificando fechas o servicios reduce el riesgo de caída a solo el 15%.

Hipótesis 4: El factor precio

```python
%sql
SELECT 
    hotel,
    COUNT(*) AS Total_Canceladas,
    AVG(adr) AS Ticket_Promedio_Perdido
FROM cursodataacademy.proyecto_dataandino.hotel_bookings
WHERE is_canceled = 1
GROUP BY hotel;
```

Conclusión: El precio promedio cancelado es idéntico en ambos hoteles. Sin embargo, el City Hotel triplica en volumen de caídas al resort, generando la mayor fuga real de capital.
